```python
import matplotlib.pyplot as mb
import seaborn as sns
```

```python
# Churn Distribution

sns.countplot(data = df,x = 'Churn')
mb.title('Customer Churn Distribution')
mb.xlabel('Churn(0 = Stayed,1 =  Churned)')
mb.ylabel('Number of Customers')
mb.show()
```

```python
# Churn by Gender

sns.countplot(data = df,x ='Gender',hue ='Churn')
mb.title('Churn by Gender')
mb.show()
```

```python
# churn by age group

import matplotlib.pyplot as mb
import seaborn as sns
mb.figure(figsize = (8,4))
sns.countplot(data = df,x ='Age Group',hue ='Churn')
mb.title('Churn by Age Group')
mb.xticks(rotation = 45)
mb.show()
```

```python
# churn by married status

sns.countplot(data = df, x ='Married',hue = 'Churn')
mb.title('Churn by Married Status')
mb.show()
```

```python
# Tenure VS Churn

mb.figure(figsize=(8,4))
sns.boxplot(data= df, x='Churn',y= 'Tenure in Months')
mb.title('Tenure Distribution by Churn')
mb.xlabel('Churn(0 = Stayed,1= churned)')
mb.show()
```

```python
# Monthly Charge Vs Churn

mb.figure(figsize=(8,4))
sns.boxplot(data = df,x = 'Churn',y = 'Monthly Charge')
mb.title('Monthly Charge by Churn')
mb.xlabel('Churn(0 = stayed,1 = Churned)') 
mb.show()
```

```python
# Top 10 churn reasons

reasons = df[df['Customer Status'] == 'Churned']['Churn Reason'].value_counts().head(10) 
mb.figure(figsize=(8,6))
sns.barplot(x = reasons.values,y = reasons.index)
mb.title('Top 10 Reasons for Customer Churn')
mb.xlabel('Number of Customers')
mb.ylabel('Churn Reason')
mb.show()
```

```python
# Churn Category

Category = df[df['Customer Status'] == 'Churned']['Churn Category'].value_counts() 
mb.figure(figsize=(7,6))
sns.barplot(x = Category.index,y = Category.values)
mb.title('Customer Churn by Category')
mb.xlabel('Churn Category')
mb.ylabel('Number of Customers')
mb.xticks(rotation =45)
mb.show()
```
